# Qt–Ft agglomeration: animation

Renders one replica of a finished run as a video (H.264, 3840 × 2160 at the default `scale = 2`): the
whole box in 3D and from the front, a fixed central magnification, the average cluster size over time and the calcium protocol.
One video frame per saved `traj.dump` frame. Output: `<run>/rep_XX/animation/animation.mp4` (and optional
stills `still_<t>s.png`). Needs ffmpeg in the `lammps` env (`conda install -c conda-forge ffmpeg`);
the stills work without it. LRZ runs: download the results first with `fetch(run_dir)` (control notebook).

```python
from qtft import animate

run = "675Qt_2700Ft_L3000_kQQ7.5_kFF3_kQF4.3_K4.3_kon1.86e-06_dt500ns_ca10-30-90s_periodic"   # run folder name (or path)
replica = 1                                    # rep_01, ...
fps = 30                                       # video frames per second
title = "Reversible assembly of protein cages"
scale = 2                                      # frame size 1920 × 1080 × scale: 1 = full HD, 2 = 4K
stills = ()                                    # optional: production times [s] also saved as PNG, e.g. (5, 30, 100)
```

```python
animate(run, replica=replica, fps=fps, title=title, scale=scale, stills=stills)
```
